```python
try:
    f = open('a2.txt')
    if f.name==('a2.txt'):
        raise Exception
except  FileNotFoundError :
    print("file does not exist")
except Exception as e:
    print(e)
else:
    print(f.read())
    f.close()
finally:
    print("run completely")
    
```

```python
try:
    a=int(input("Enter the number 1:"))
    b=int(input("Enter the number 2:"))
    print(a/b)
except ZeroDivisionError:
    print(f"Cannot divisible by zero")
except ValueError:
    print(f"Invalid datatype")
finally:
    print(f"run Completely")
```
