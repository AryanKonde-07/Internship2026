```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array([10,23,29,34,39,45,50])
y=np.array([12,21,31,43,48,55,58])
z=np.array([56,50,44,37,26,20,16])
```

```python
plt.ylabel("y value")
plt.title("Line Chart:x vs y")
plt.plot(x,y,marker='o',mfc='red',ls='--');
plt.grid();

plt.xlabel("x value")
plt.ylabel("y value")
plt.title("Line Chart:x vs y")
plt.plot(x,z,marker='o',mfc='red',ls='--');
plt.grid();
```

```python
plt.figure(figsize=(16,9))
plt.subplot(2,2,1)
plt.xlabel("x value")
plt.ylabel("y value")
plt.title("Line Chart:x vs y")
plt.plot(x,y,marker='o',mfc='red',ls='--');
plt.grid();

plt.subplot(2,2,2)
plt.xlabel("x value")
plt.ylabel("y value")
plt.title("Line Chart:x vs z")
plt.plot(x,z,marker='o',mfc='red',ls='--');
plt.grid();
```

```python
#load ->Social_Network_Ads.csv
#plot bar graph showing the count of how many people purchsed and how many did not
#purchase column =>0->purchased,1->not purchased
```

```python
df=pd.read_csv("Social_Network_Ads.csv")
df
```

```python

counts = df['Purchased'].value_counts()

purchased = counts.get(0, 0)
not_purchased = counts.get(1, 0)



plt.bar(['Purchased', 'Not Purchased'],
        [purchased, not_purchased])

plt.title("Purchase Status vs Not Purchased")
plt.xlabel("Purchased Status")
plt.ylabel("count")
plt.show()
```

```python
df=pd.read_csv('Social_Network_Ads.csv')# Another Solutionnn
purchased_counts=df['Purchased'].value_counts()

bars=purchased_counts.plot(kind='bar',color='cyan',edgecolor='red')
plt.title('Distribution of Purchase Behavior')
plt.xlabel('Purchase Status')
plt.ylabel('Number of People')
plt.xticks(rotation=50)
plt.show();
```
