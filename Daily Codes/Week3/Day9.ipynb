```python
class Person:
    def __init__(self,name,age,ht):
        
        self.name=name
        self.__age=age     # to give a privte variable we use ("__")
        self.ht=ht
    def show_info(self):
        return f"name-.{self.name},age->{self.__age},height->{self.ht}"
    def age(self):
        print(self.__age)
        
```

```python
p1=Person("Aryan",18,6.11)
```

```python
p1.show_info()
```

```python
p1.age
```

```python
class Employee:
    company="Wipro"
    def __init__(self,name,monthly_salary,job_designation):
        self.name=name
        self.__monthly_salary=monthly_salary
        self.job_designation=job_designation
    def calculate_annaul(self):
        print(f"Monthly Salary=>",(self.__monthly_salary)*12)
    def show_info(self):
        print (f" company->{self.company},name->{self.name},monthly_salary->{self.__monthly_salary},job_designation->{self.job_designation}")
```

```python
e1=Employee("Aryan",25000,"AI Engineer")
e2=Employee("Swaraj",15000,"Web-Designer")
e3=Employee("Atharav",35000,"HR")
```

```python
e1.show_info()
```
