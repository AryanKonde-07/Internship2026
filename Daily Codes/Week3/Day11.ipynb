```python
l1 = [1, 2, 3, 2, 1] 
l2 = ['a', 'b', 'c', 'b', 'd']
l3=list1.copy()
l4=list2.copy()
l1.reverse()
l2.reverse()
if l1==l3:
    print ("Palindrome")
else:
    print("Not Palindrome")

if l2==l4:
    print("Palindrome")
else:
    print("Not Palindrome")
```
