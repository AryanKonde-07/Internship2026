# Inheritance

```python
class Person:
    insti="MITU"
    def __init__(self,age,ht):
        self.age=age
        self.ht=ht

    def show(self):
        print(f"age=>{self.age}, ht=>{self.ht}, institute=>{self.insti}")     
```

```python
class Ravi(Person):
    def __init__(self,age,ht,sal):
        super().__init__(age,ht)
        self.sal=sal
    def show_sal(self):
        return f"Salary is=>{self.sal}"
```

```python
r1=Ravi(22,5,5000)
```

```python
r1.show_sal()
```

```python
r1.show()
```

```python
class Menuapp:
    def __init__(self):
        self.itemlist=[]
    def show_menu(self):
        print ("Menu\n1.add_items \n2.view_items \n3.exit")
    def add_items(self):
        item=input("Enter item name:")
        self.itemlist.append(item)

    def view_items(self):
        print(self.itemlist)
        
    def run(self):
        while True:
            self.show_menu()
            ch = int(input("enter the choice"))
            if ch==1:
                self.add_items()
            elif ch==2:
                self.view_items()
            elif ch==3:
                break
            else:
                return ("Invalid Choice")
```

```python
m=Menuapp()
```

```python
m.run()
```

```python
# check a manu driven app for banking simulation
#check balance ,deposit money ,withdraw money(u cann add more things as u see fit to make the app better)
```

```python
class MenuDriven:
    def __init__(self):
        self.balance=50000
    def show_menu(self):
        print ("Menu\n1.Withdraw \n2.Deposite \n3.Check Balance")
    def withdraw(self):
        amount = int(input("Enter the amount to withdraw: "))
        self.balance -= amount
        print(f"Withdraw amount=>{amount}")
        print(f"Available Balance=>{self.balance}")

    def deposite(self):
        amount = int(input("Enter the amount to deposit: "))
        self.balance += amount
        print(f"Deposit amount=>{amount}")
        print(f"Available Balance=>{self.balance}")

    def CheckBal(self):
        print(f"Available Balance=>{self.balance}")

    def run(self):
        while True:
            self.show_menu()
            ch = int(input("Enter the choice: "))

            if ch == 1:
                self.withdraw()
            elif ch == 2:
                self.deposite()
            elif ch == 3:
                self.CheckBal()
            else:
                print("Invalid Choice")
                break
```

```python
m=MenuDriven()
```

```python
m.run()
```
