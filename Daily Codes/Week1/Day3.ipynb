```python
"""
menu Driven
take a string from user
menu->1 upper,2 lowercase,3 replace a particular part of the string
for third option it should which part to replce with what
""";
```

```python
print("Menu Driven")
string=input("Enter the String:")
print("Menu->\n1.uppercase\n2.lowercase\n3.repalce")
ch = int(input("Select the choice:"))

         
if ch==1:
    print(string.upper())
elif ch==2:
    print(string.lower())
elif ch==3:
    ch1=input("Enter the replace string")
    ch2=input("Enter with what to replace")
    print(string.replace(ch1,ch2))
else:
            print("Invalid Choice")
        
```
