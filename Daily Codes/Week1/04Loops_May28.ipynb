# Loop Control Statements ->Break,Continue

```python
for i in range (1,11):
    print(i)
```

```python
for i in range(1,11):
    if i==5:
        break
    print(i)
    print("Ok")
```

```python
for i in range (1,11):
    print(i)
    print('ok')
    if i==5:
     break
```

```python
for i in range(1,11):
    if i ==5:
        continue
    print(i)
```

```python
for i in range(1,11):
    print(i)
    if i==5:
        continue
    print(i)
    print('ok')
```
