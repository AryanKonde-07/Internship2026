```python
#1. Employee Management System

'''Create a base class Employee with:

emp_id
name
salary

Create derived classes:

Manager
Developer

Requirements:

Override a method calculate_bonus().
Manager gets 20% bonus.
Developer gets 10% bonus.

Demonstrate inheritance and polymorphism.''';
```

```python
class Employee:
    def __init__(self, emp_id, name, salary):
        self.emp_id = emp_id
        self.name = name
        self.salary = salary


class Manager(Employee):
    def calculate_bonus(self):
        return self.salary * 0.20


class Developer(Employee):
    def calculate_bonus(self):
        return self.salary * 0.10
```

```python
m1 = Manager(101, "Rahul", 500000)
d1 = Developer(102, "Priya", 400000)

print("Manager Bonus:", m1.calculate_bonus())
print("Developer Bonus:", d1.calculate_bonus())
```

```python
#2.Library Management System

'''Create classes:

Book
Member
Library

Requirements:

Add books.
Issue books to members.
Return books.
Prevent issuing an already issued book.

Use object composition.''';
```

```python
# Book class
class Book:
    def __init__(self, title):
        self.title = title
        self.issued = False


# Member class
class Member:
    def __init__(self, name):
        self.name = name


# Library class
class Library:
    def __init__(self):
        self.books = []  # List to store Book objects

    def add_book(self, book):
        self.books.append(book)
        print(book.title, "added to library")

    def issue_book(self, book, member):
        if book.issued:
            print(book.title, "is already issued")
        else:
            book.issued = True
            print(book.title, "issued to", member.name)

    def return_book(self, book):
        if book.issued:
            book.issued = False
            print(book.title, "returned successfully")
        else:
            print(book.title, "was not issued")


```

```python
b1 = Book("Python")
b2 = Book("Data Science")

lib = Library()
m1 = Member("Rahul")
```

```python
lib.add_book(b1)
lib.add_book(b2)
```

```python
lib.issue_book(b1, m1)
```

```python
lib.issue_book(b1, m1)
```

```python
lib.return_book(b1)
```

```python
lib.issue_book(b1, m1)
```

```python
#3.Student Result Processing

'''Create a class Student with:

roll_no
name
marks (list)

Methods:

calculate_percentage()
calculate_grade()

Requirements:

Use private attributes.
Provide getter/setter methods.
Grade rules:
A: ≥ 80
B: ≥ 60
C: ≥ 40
Fail: < 40
''';
```

```python
class Student:
         # Getter methods
    def get_roll_no(self):
        return self.__roll_no

    def get_name(self):
        return self.__name

    def get_marks(self):
        return self.__marks
        # Setter methods
    def set_roll_no(self, roll_no):
        self.__roll_no = roll_no

    def set_name(self, name):
        self.__name = name

    def set_marks(self, marks):
        self.__marks = marks


    # Calculate percentage
    def calculate_percentage(self):
        return sum(self.__marks) / len(self.__marks)
        # Calculate grade
    def calculate_grade(self):
        percentage = self.calculate_percentage()

        if percentage >= 80:
            return "A"
        elif percentage >= 60:
            return "B"
        elif percentage >= 40:
            return "C"
        else:
            return "Fail"

```
