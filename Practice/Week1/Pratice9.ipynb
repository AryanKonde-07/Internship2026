```python
'''Create a Student class that stores marks for 5 subjects and has a method to calculate the percentage.''';
```

```python
class Student:
    def __init__(self, name, marks):
        """
        Initialize the student with a name and a list of 5 subject marks.
        """
        self.name = name
        self.marks = marks # Expecting a list of 5 numbers

    def calculate_percentage(self):
        """
        Calculate and return the percentage for the 5 subjects.
        Assuming each subject is out of 100.
        """
        total_marks = sum(self.marks)
        percentage = (total_marks / 500) * 100
        return percentage

    
```

```python
student1 = Student("Rahul", [85, 90, 78, 92, 88])
```

```python
print(f"Student Name: {student1.name}")
print(f"Marks: {student1.marks}")
print(f"Percentage: {student1.calculate_percentage():.2f}%")
    
```
